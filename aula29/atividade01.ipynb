```python
import pandas as pd
import numpy as np
```

```python
ENDERECO_DADOS = 'https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv'

df_estupro = pd.read_csv(ENDERECO_DADOS, sep=';' , encoding= 'iso-8859-1')
```

```python
df_estupro = df_estupro[['cisp','regiao', 'munic', 'estupro']]
correcao_acentuacao = df_estupro['regiao'].str.startswith('Grande Niter', na=False)

df_estupro.loc[correcao_acentuacao, 'regiao'] = 'Grande Niterói'
print(df_estupro['regiao'].unique())
```

```python
df_estupro = df_estupro.groupby(['cisp', 'regiao', 'munic'], as_index=False)['estupro'].sum()
print(df_estupro.head(30))
```

```python
array_estupro = np.array(df_estupro['estupro'])

media = np.mean(array_estupro)
mediana = np.median(array_estupro)

maximo = np.max(array_estupro)
minimo = np.min(array_estupro)
total = np.sum(array_estupro)

q1 = np.percentile(array_estupro,25)
q3 = np.percentile(array_estupro,75)

print(f'Média: {media}')
print(f'Mediana: {mediana}')

print(f'Máximo: {maximo:.2f}')
print(f'Mínimo:  {minimo:.2f}')
print(f'Total:  {total:.2f}')

print( f'Q1: {q1:.2f}' )
print( f'Q3: {q3:.2f}' )
```

```python
# Criando uma série com os maiores
df_maiores = df_estupro[df_estupro['estupro'] >=q3].copy()
df_maiores['flag'] = 'mais'
display(df_maiores)

df_menores = df_estupro[df_estupro['estupro'] <=q1].copy()
df_menores['flag'] = 'menos'
display(df_menores)
```

```python
df_estupro_flags = pd.concat([df_maiores,df_menores], ignore_index=True)
```

```python
# Exportando os dados utf-8-sig p/ ler o powerbi
df_estupro_flags.to_csv('estupro_flag.csv', index=False, encoding='utf-8-sig')
```
